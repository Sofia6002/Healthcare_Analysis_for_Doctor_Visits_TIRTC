### Univariate Plot 1: Distribution of Gender

Let's visualize the distribution of the 'gender' column using a count plot.

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='gender', data=df, hue='gender', palette='viridis', legend=False)
plt.title('Distribution of Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

### Univariate Plot 2: Distribution of Age

Next, let's look at the distribution of the 'age' column using a histogram with a Kernel Density Estimate (KDE) to see its shape.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['age'], kde=True, bins=20)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

### Univariate Plot 3: Distribution of Number of Visits

Finally, let's visualize the distribution of the 'visits' column, which represents the number of doctor visits.

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='visits', data=df, hue='visits', palette='viridis', legend=False)
plt.title('Distribution of Number of Doctor Visits')
plt.xlabel('Number of Visits')
plt.ylabel('Count')
plt.show()
```

```python
gender_analysis = (
    df.groupby("gender")["visits"]
    .agg(["count", "mean", "median"])
    .reset_index()
)

display(gender_analysis)
```

```python
sns.barplot(
    data=gender_analysis,
    x="gender",
    y="mean",
    color="teal"
)

plt.title("Average Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Mean recorded visits")
plt.show()
```
